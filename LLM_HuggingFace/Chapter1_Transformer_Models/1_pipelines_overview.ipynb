```python
from transformers import pipeline
```

```python
classifier = pipeline("sentiment-analysis")
classifier("I love using transformers for natural language processing tasks!")
```

**Several text at one time**

```python
classifier(
    ["I have been waiting to start this topic, i am so excited", "I hate the fact that i was sick in the important moment"]
)
```

**Labels Pipeline** ( zero-shot-classification)

```python

classifier = pipeline("zero-shot-classification")
```

```python
classifier(
    "This is the notebook about transformer library",
    candidate_labels=['education','politics','informative'],
)
```

**Text Generation Pipeline** ( text-generation)

```python
generator = pipeline("text-generation")
generator("The transformers library is a powerful tool for natural language  ")
```

**Text Generation pipeline with specific model and arguments**

```python
generator = pipeline("text-generation", model='distilgpt2')

generator(
    "In this notebook, we will learn about transformers and",
    max_length= 30,
    num_return_sequences=2,
)
```
