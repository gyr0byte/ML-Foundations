```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

**<h3>Loading the dataset</h3>**

```python
df = pd.read_csv("train.txt", sep=';', header=None, names=['text','emotion'])
```

```python
df.head()
```

**<h3>Data Cleaning</h3>**

```python
df.isnull().sum()
```

converting the output variables into numbers

```python
unique_emotions = df['emotion'].unique()
emotion_numbers = {}
i = 0
for emo in unique_emotions:
    emotion_numbers[emo] = i
    i += 1
    
df['emotion'] = df['emotion'].map(emotion_numbers)
```

```python
df.head()
```

changing all the text into the lower case

```python
df['text'] = df['text'].apply(lambda x : x.lower())
```

removing the punctuations

```python
import string

def remove_punc(txt):
    return txt.translate(str.maketrans('','',string.punctuation))
```

```python
df['text'] = df['text'].apply(remove_punc)
```

removing the numbers

```python
def remove_numbers(txt):
    new = ""
    for i in txt:
        if not i.isdigit():
            new = new + i
    return new


df['text'] = df['text'].apply(remove_numbers)
```

removing the emojis

```python
def remove_emojis(txt):
    new = ""
    for i in txt:
        if i.isascii():
            new += i
    return new

df['text'] = df['text'].apply(remove_emojis)
```

removing stop wrods using NLTK

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt')
nltk.download('stopwords')
```

```python
stop_words = set(stopwords.words("english"))
```

```python
df.loc[1]['text']
```

```python
# tokenization
def remove(txt):
    words = txt.split()
    cleaned = []
    for i in words:
        if not i in stop_words:
            cleaned.append(i)
    return " ".join(cleaned)
```

```python
df['text'] = df['text'].apply(remove)
```

```python
df.loc[1]['text']
```

**Spliting the training and testing dataset**

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(df['text'],df['emotion'],test_size=0.2, random_state=42)
```

vectorize the y_train and y_test

```python
from sklearn.feature_extraction.text import TfidfVectorizer, CountVectorizer
```

**<h3>Bag of words</h3>**

```python
bow_vectorizer = CountVectorizer()
```

```python
X_train_bow = bow_vectorizer.fit_transform(X_train)
X_test_bow = bow_vectorizer.transform(X_test)
```

**Using a ML model (Naive Bayes)**

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
```

```python
nb_model = MultinomialNB()
```

```python
nb_model.fit(X_train_bow,y_train)
```

```python
pred_nb = nb_model.predict(X_test_bow)
```

```python
print(accuracy_score(y_test,pred_nb))
```
