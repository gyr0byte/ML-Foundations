# Categorical Plots with Seaborn

This notebook explores categorical visualizations using the `tips` dataset:

- Import Matplotlib and Seaborn.
- Load the `tips` dataset and preview it.
- Create a count plot by `sex` with `smoker` as hue.
- Build bar plots of `total_bill` by `sex`, including a custom `estimator` (sum).
- Compare distributions with box plots (column and label forms).
- Visualize distributions with violin, strip, and swarm plots.

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset('tips')
df
```

**<h4>Count Plot</h4>**

```python
sns.countplot(x = df['sex'], hue = df['smoker'])
```

**<h4>Bar Plot</h4>**

```python
sns.barplot(x=df['sex'], y = df['total_bill'])
```

**estimator**

```python
import numpy as np
sns.barplot(x=df['sex'], y = df['total_bill'], estimator=np.sum)
```

**<h4>Box Plot</h4>**

```python
sns.boxplot(x= df['tip'], y = df['day'])
```

```python
sns.boxplot(x= 'tip', y = 'day',data=df)
```

**<h4>Violin Plot</h4>**

```python
sns.violinplot(x= 'day', y = 'total_bill', data=df)
```

**<h4>Strip Plot</h4>**

```python
sns.stripplot(x= 'day', y = 'total_bill', data=df)
```

**<h4>Swarm Plot</h4>**

```python
sns.swarmplot(x= 'day', y = 'total_bill', data=df)
```

**combining** violin and swarm plots

```python
sns.violinplot(x= 'tip', y = 'day', data=df)
sns.swarmplot(x= 'tip', y = 'day', data=df)
```
