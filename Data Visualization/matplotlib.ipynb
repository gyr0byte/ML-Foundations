```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

**<h3>Matplotlib Basics</h3>**

```python
x = np.linspace(0, 5, 11)
x
y = x ** 2
y
```

```python
plt.plot(x,y)
plt.title("Line Graph")
plt.xlabel("X axis")
plt.ylabel("Y axis")
```

**<h3>Subplots in Matplotlib</h3>**

```python
plt.subplot(2,2,1)
plt.plot(x,y**2)
plt.subplot(2,2,2)
plt.plot(x**2,x**2)
plt.subplot(2,2,3)
plt.plot(y,x)
plt.subplot(2,2,4)
plt.plot(x,y)
```

**<h3>Object Oriented way of plotting</h3>**

```python
fig = plt.figure()
axis1 = fig.add_axes([0.1,0.1,0.8,0.8]) # type: ignore
axis1.plot(x,y)
axis2 = fig.add_axes([0.1,0.1,0.8,0.8]) # type: ignore
axis2.plot(x,y)
```
