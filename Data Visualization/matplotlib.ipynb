# Matplotlib Overview

This notebook covers core Matplotlib plotting patterns:

- Import NumPy, Pandas, and Matplotlib.
- Create a simple line plot with titles and axis labels.
- Build a 2x2 subplot grid with multiple line plots.
- Use the object-oriented API with `figure()` and `add_axes()`.
- Demonstrate common plot types: scatter, histogram, and box plot.
- Save a plot to a file with `savefig()`.
- Load and display images using `matplotlib.image` and `imshow()`.
- Apply basic styling with color options.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

**<h3>Matplotlib Basics</h3>**

```python
x = np.linspace(0, 5, 11)
x
y = x ** 2
y
```

```python
plt.plot(x,y)
plt.title("Line Graph")
plt.xlabel("X axis")
plt.ylabel("Y axis")
```

**<h3>Subplots in Matplotlib</h3>**

```python
plt.subplot(2,2,1)
plt.plot(x,y**2)
plt.subplot(2,2,2)
plt.plot(x**2,x**2)
plt.subplot(2,2,3)
plt.plot(y,x)
plt.subplot(2,2,4)
plt.plot(x,y)
```

**<h3>Object Oriented way of plotting</h3>**

```python
fig = plt.figure(figsize=(10,8), dpi=100)
axis1 = fig.add_axes([0.1,0.1,1,1]) # type: ignore
axis1.plot(x,y)
axis2 = fig.add_axes([0.3,0.1,0.3,0.3]) # type: ignore
axis2.plot(x,y)
```

**<h3>Types of plots</h3>**

**scatter plot**

```python
plt.scatter(x,y)
```

**Histogram**

```python
from random import sample
data = sample(range(1,1000), 100)
plt.hist(data)
```

**Box plot**

```python
plt.boxplot(x)
```

**<h3>Saving plots</h3>**

```python
fig = plt.figure()
ax = fig.add_axes([0.1,0.1,0.8,0.8]) #type: ignore
ax.plot(x,y)
plt.savefig("basic_plot.png")
```
