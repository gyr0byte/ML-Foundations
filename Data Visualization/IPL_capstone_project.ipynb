**<h4>Load libraries and dataset</h4>**

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
df = pd.read_csv('IPL.csv')

df.head()
```

**1. Which team won the most matches?**

```python
match_wins = df['match_winner'].value_counts()
sns.barplot(y = match_wins.index, x =match_wins.values, palette="viridis")
plt.title("Most match win by team")
```

**2. Toss Decision Trends**

```python
sns.countplot(x = df['toss_decision'])
plt.title("Toss Decision Trends")
```

**3. Toss Winners vs Match Winners**

```python
count = df[df['toss_winner'] == df['match_winner']]['match_id'].count()
percentage = (count * 100)/df.shape[0]
percentage.round(2) # type: ignore
```

**4. How do teams win?** (runs vs wicket)

```python
sns.countplot(df['won_by']) # type: ignore
```

**<h4>Key Player Performances</h4>**

**1. Most "Player of the match" Awards**

```python
count = df['player_of_the_match'].value_counts().head(10)
count
```

```python
sns.barplot(x = count.values, y = count.index, palette="mako")
plt.title("Top 10 players with man of the match")
```

**2. Top 2 Scorers**

```python
high = df.groupby('top_scorer')['highscore'].sum().sort_values(ascending=False).head(2)
high
```

```python
high.plot(kind = 'barh')
```

**3. 10 Best Bowling Figures**

```python
df['highest_wickets'] = df['best_bowling_figure'].apply(lambda x:x.split('--')[0])
df['highest_wickets'] = df['highest_wickets'].astype(int)
top_bowlers = df.groupby('best_bowling')['highest_wickets'].sum().sort_values(ascending=False).head(10)
top_bowlers
```

```python
top_bowlers.plot(kind='barh')
```

**<h4>Venue Analysis</h4>**

Most matches played in which venue

```python
venue_counts = df['venue'].value_counts()
venue_counts
```

```python
sns.barplot(x = venue_counts.values, y = venue_counts.index, palette="viridis")
plt.title("Most matches played in which venue")
```

**<h4>Custom Question and Insights</h4>**

Q1: Who won the highest margin by runs?

```python
df[df['won_by'] == 'Runs'].sort_values(by='margin',ascending=False).head(1)[['match_winner','margin']]
```

Q2: Which player had the highest individual score?

```python
df[df['highscore'] == df['highscore'].max()][['top_scorer','highscore']]
```

Q3: Which bowler had the best bowling figures?

```python
df[df['highest_wickets'] == df['highest_wickets'].max()][['best_bowling', 'best_bowling_figure']]
```
