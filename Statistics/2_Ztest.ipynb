```python
import numpy as np
from scipy.stats import norm
```

**<h5>Given<h5>**

```python
sample = [172, 174, 168, 169, 171, 173, 175, 170, 169, 172]
population_mean = 170
population_std = 3
sample_mean = np.mean(sample)
n = len(sample) # sample size is not 30 just for the example only otherwise it is not right to 
                # use sample size less than 30 for a z test
```

**<h5>Z score calculation<h5>**

```python
z_score = (sample_mean - population_mean) / (population_std / np.sqrt(n))
z_score
```

**<h5>P value calculation<h5>**

```python
p_value = 2 * (1- norm.cdf(abs(z_score)))
p_value
```

```python
alpha = 0.05

if p_value < alpha:
    print("I will reject my null hypothesis")
else:
    print("I will accept the null hypothesis")
```
