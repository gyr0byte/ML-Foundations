```python
import numpy as np
arr = np.array([2, 3, 4, 6, 7, 8, 9, 12, 13, 16, 17, 23, 25, 27, 34, 37, 201])
arr
```

**<h5>Calculating and storing percentile</h5>**

```python
Q1 = np.percentile(arr,25)
Q3 = np.percentile(arr,75)
```

**<h5> Calculating IQR</h5>**

```python
IQR = Q3 - Q1
IQR
```

**<h5>Calculating upper fence and lower fence</h5>**

```python
upper_fence = Q3 + (1.5*IQR)
lower_fence = Q1 - (1.5*IQR)
```

**<h5> Removing the outlier element</h5>**

```python
l = []
for i in arr:
    if i <= upper_fence and i >= lower_fence:
        l.append(i)
arr2 = np.array(l)
```

```python
print(arr)
print(arr2)
```
