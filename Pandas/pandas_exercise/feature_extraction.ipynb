extract episode function

```python
def extract_episodes(txt):
    check = False
    data = ""
    for i in txt:
        if i == ")":
            break
        if check == True:
            data += i
        if i == "(":
            check = True
    return data
```

extracting episode using apply

```python
df['Episodes'] = df['Title'].apply(extract_episodes) 
```

removing the eps string with empty string

```python
df["Episodes"] = df["Episodes"].str.replace(" eps", "")
```

converting episode from string to numbers

```python
df["Episodes"] = df["Episodes"].astype(int)
```

**Make new column for time stamp**

```python
df.loc[0]['Title']
```

```python
def extract_time(txt):
    check = False
    data = ""
    for i in range(len(txt)):
        if txt[i] == ")":
            for j in range(i + 1, i+20):
                data += txt[j]
            return data
```

```python
df["Total Time"] = df["Title"].apply(extract_time)
```

```python
df
```

```python
def extract_months(value):
    try:
        start, end = value.split(' - ')

        start_date = pd.to_datetime(start, format='%b %Y')
        end_date = pd.to_datetime(end, format='%b %Y')

        months = (
            (end_date.year - start_date.year) * 12 +
            (end_date.month - start_date.month)
        )

        return months

    except:
        return None
```

```python
df['Months'] = df['Total Time'].apply(extract_months)
```
